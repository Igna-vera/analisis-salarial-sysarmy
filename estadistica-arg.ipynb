```python
# Calculo cuartiles del salario bruto
q1 = df_limpio[col_bruto].quantile(0.25)
q3 = df_limpio[col_bruto].quantile(0.75)

# Calculo  rango intercuartílico
iqr = q3 - q1

# Calculamos los límites
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

# Identifico posibles valores raros
outliers = df_limpio[
    (df_limpio[col_bruto] < limite_inferior) |
    (df_limpio[col_bruto] > limite_superior)
]

# Resultados
print("Primer cuartil (Q1):", round(q1, 2))
print("Tercer cuartil (Q3):", round(q3, 2))
print("Rango intercuartílico (IQR):", round(iqr, 2))

print("\nLímite inferior:", round(limite_inferior, 2))
print("Límite superior:", round(limite_superior, 2))

print("\nCantidad de posibles outliers:", len(outliers))

print(
    "Porcentaje de posibles outliers:",
    round(len(outliers) / len(df_limpio) * 100, 2),
    "%"
)

# 10 salarios más altos identificados
display(
    outliers[[col_bruto, "seniority"]]
    .sort_values(by=col_bruto, ascending=False)
    .head(10)
)
```

```python
# Slarios por seniority
resumen_outliers = []

for nivel in ["Junior", "Semi-Senior", "Senior"]:

    # Filtro los salarios del nivel actual
    salarios = df_limpio.loc[
        df_limpio["seniority"] == nivel,
        col_bruto
    ]

    # Calculo cuartiles
    q1 = salarios.quantile(0.25)
    q3 = salarios.quantile(0.75)
    iqr = q3 - q1

    # Calculo el límite superior
    limite_superior = q3 + 1.5 * iqr

    # Outliers
    cantidad_outliers = (salarios > limite_superior).sum()

    # Guardado los resultados
    resumen_outliers.append({
        "seniority": nivel,
        "cantidad_respuestas": len(salarios),
        "limite_superior": limite_superior,
        "cantidad_outliers": cantidad_outliers,
        "porcentaje_outliers": (
            cantidad_outliers / len(salarios) * 100
        )
    })

# Tabla resultados
resumen_outliers = pd.DataFrame(resumen_outliers)

# Valores
resumen_outliers = resumen_outliers.round(2)

display(resumen_outliers)
```

```python
# Identifico los dos grupos de posibles inconsistencias

neto_superior = df_limpio[col_neto] > df_limpio[col_bruto]

neto_muy_inferior = (
    (df_limpio["relacion_neto_bruto"] < 0.10) &
    (df_limpio[col_neto] > 0)
)

print("Salario neto superior al bruto:", neto_superior.sum())

print("Salario neto inferior al 10% del bruto:", neto_muy_inferior.sum())

print(
    "Registros que cumplen ambas condiciones:",
    (neto_superior & neto_muy_inferior).sum()
)

print(
    "Total de registros diferentes para revisar:",
    (neto_superior | neto_muy_inferior).sum()
)

# Reviso los valores de la relación neto/bruto
print("\nDistribución de la relación neto/bruto:")
display(
    df_limpio["relacion_neto_bruto"]
    .describe()
    .round(3)
)
```

```python
# Clasifico los registros que necesitan revisión

df_revision = df_limpio.loc[
    neto_superior | neto_muy_inferior
].copy()

df_revision["tipo_inconsistencia"] = np.where(
    neto_superior.loc[df_revision.index],
    "Neto superior al bruto",
    "Neto inferior al 10% del bruto"
)

# Comparo las inconsistencias con la modalidad de cobro en dólares

print("Modalidad de cobro en los registros observados:")
display(
    df_revision["pagos_en_dolares"]
    .value_counts(dropna=False)
)

print("\nSueldo dolarizado:")
display(
    df_revision["sueldo_dolarizado"]
    .value_counts(dropna=False)
)

print("\nModalidad de cobro según tipo de inconsistencia:")
display(
    pd.crosstab(
        df_revision["tipo_inconsistencia"],
        df_revision["pagos_en_dolares"],
        margins=True,
        dropna=False
    )
)
```

```python
# Clasifico los registros con posibles inconsistencias salariales

df_limpio["tipo_inconsistencia_neto"] = pd.NA

df_limpio.loc[
    neto_superior,
    "tipo_inconsistencia_neto"
] = "Neto superior al bruto"

df_limpio.loc[
    neto_muy_inferior,
    "tipo_inconsistencia_neto"
] = "Neto inferior al 10% del bruto"

# Reviso la clasificación
print("Clasificación de posibles inconsistencias:")
display(
    df_limpio["tipo_inconsistencia_neto"]
    .value_counts()
    .to_frame("cantidad")
)

# Conservo una tabla separada para futuras revisiones
df_neto_revision = df_limpio[
    df_limpio["tipo_inconsistencia_neto"].notna()
].copy()

print(
    "\nRegistros conservados para revisión:",
    len(df_neto_revision)
)
```
